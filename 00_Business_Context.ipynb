## 1. BỐI CẢNH KINH DOANH


Danny – một doanh nhân trẻ đam mê ẩm thực – đã khai trương **Pizza Runner**, một startup pizza kết hợp bếp nướng thủ công và đội ngũ giao hàng riêng. Khởi đầu chỉ là một gian bếp nhỏ, vài chiếc xe máy và một cuốn sổ tay ghi đơn. Nhưng Danny không muốn dừng lại ở đó: anh khao khát thấu hiểu khách hàng, tối ưu từng khâu vận hành và mở rộng quy mô một cách thông minh.

**Dữ liệu hiện tại** chính là “tài sản số” đầu tiên của Danny – được thu thập qua ứng dụng đặt hàng tự phát triển. Dù còn thô sơ, nó ghi lại toàn bộ hành trình: từ lúc khách đặt món, công thức chế biến, cho đến khi đơn hàng được trao tận tay. Dự án phân tích này ra đời nhằm giải đáp những câu hỏi sống còn, giúp Danny vượt qua từng cột mốc tăng trưởng.

> **Tầm nhìn phân tích:** Biến dữ liệu thô thành “trợ lý CEO” – trao cho Danny khả năng ra quyết định chính xác, giảm lãng phí và tạo ra trải nghiệm khách hàng đáng nhớ.

## 2.  LỘ TRÌNH PHÁT TRIỂN  CÁC GIAI ĐOẠN PHÂN TÍCH


Hành trình của Pizza Runner được chia thành **4 giai đoạn**, mỗi giai đoạn gắn với một câu hỏi lớn mà Danny phải trả lời. Các file phân tích A, B, C, D lần lượt làm rõ từng phần – và khi ghép lại, chúng tạo thành bức tranh toàn cảnh giúp Danny đi từ “sống sót” đến “tăng trưởng bền vững”.

### **2.1. Khởi nghiệp** (*A_Pizza_Metrics*)

| Câu hỏi quản trị | Các câu hỏi phân tích | Mục tiêu kinh doanh | Domain |
|------------------|------------------------|---------------------|--------|

| | – Khách hàng mua nhiều nhất | – Chăm sóc khách hàng thân thiết (Sales), thiết kế chương trình loyalty (Marketing) | Sales (Bán hàng)<br>Marketing (Tiếp thị) |
| | – Số pizza tối đa trong một đơn | – Hiểu hành vi đặt số lượng lớn để có chính sách giá sỉ hoặc combo (Sales, Menu Engineering) | Sales (Bán hàng)<br>Menu Engineering (Thiết kế thực đơn) |
| | – Lưu lượng theo giờ và ngày trong tuần | – Dự báo khung giờ cao điểm (Operations), lên ca cho nhân viên bếp và phục vụ (HR) | Operations (Vận hành)<br>HR (Nhân sự) |
| | – Tỉ lệ đơn hàng bị hủy | – Đánh giá rủi ro vận hành, tìm nguyên nhân hủy đơn để giảm thiểu (Operations) | Operations (Vận hành) |

| | – Thời gian trung bình đến lấy hàng | – Phát hiện trễ trong khâu bếp, cải thiện quy trình chuẩn bị (Operations) | Operations (Vận hành) |

| | – Tốc độ di chuyển trung bình từng runner | – Xây dựng KPI hiệu suất cá nhân, gắn với lương thưởng (HR, Operations) | HR (Nhân sự)<br>Operations (Vận hành) |
| | – Tỉ lệ giao thành công của mỗi runner | – Đánh giá chất lượng phục vụ, cơ sở để khen thưởng, đào tạo hoặc chấm dứt hợp đồng (HR) | HR (Nhân sự) |

### **2.3. Mở rộng** (*C_Ingredient_Optimisation*)

| Câu hỏi quản trị | Các câu hỏi phân tích | Mục tiêu kinh doanh | Domain |
|------------------|------------------------|---------------------|--------|
| “Làm sao bán nhiều hơn mà không đội chi phí nguyên liệu?” | – Topping được thêm nhiều nhất | – Phát hiện cơ hội upsell, tạo combo topping có tính phí tăng doanh thu (Menu Engineering, Sales) | Menu Engineering (Thiết kế thực đơn)<br>Sales (Bán hàng) |
| | – Topping bị loại bỏ nhiều nhất | – Cắt giảm lãng phí, điều chỉnh định mức nguyên liệu mua vào (Supply Chain, Finance) | Supply Chain (Chuỗi cung ứng)<br>Finance (Tài chính) |
| | – Công thức chuẩn từng pizza <br> – Danh sách nguyên liệu thực tế sau tùy chỉnh (hiển thị cho bếp) | – Chuẩn hóa hiển thị order cho bếp để giảm sai sót, dễ kiểm soát chi phí (Operations, Supply Chain) | Operations (Vận hành)<br>Supply Chain (Chuỗi cung ứng) |
| | – Tổng lượng từng nguyên liệu đã dùng trong các đơn thành công | – Tính giá vốn hàng bán thực tế, so sánh với định mức để tối ưu margin (Finance, Supply Chain) | Finance (Tài chính)<br>Supply Chain (Chuỗi cung ứng) |

### **2.4. Tăng trưởng** (*D_Pricing_and_Ratings*)

| Câu hỏi quản trị | Các câu hỏi phân tích | Mục tiêu kinh doanh | Domain |
|------------------|------------------------|---------------------|--------|
| “Định giá ra sao để có lời thật sự?” | – Doanh thu nếu giá cố định 12 dollar và 10 dollar | – Mô phỏng tác động của giá bán lên tổng doanh thu, chọn giá tối ưu (Finance, Pricing) | Finance (Tài chính)<br>Pricing (Định giá) |
| | – Doanh thu tăng thêm khi thu phí topping thêm 1 dollar | – Lượng hóa hiệu quả upsell, thiết kế chính sách giá linh hoạt cho topping (Sales, Finance) | Sales (Bán hàng)<br>Finance (Tài chính) |
| | – Thiết kế bảng đánh giá runner (1-5 sao) <br> – Bảng tổng hợp đơn hàng – giao hàng – đánh giá | – Xây dựng cơ chế phản hồi khách hàng để đảm bảo chất lượng dịch vụ, tác động đến lòng trung thành (Marketing, Operations) | Marketing (Tiếp thị)<br>Operations (Vận hành) |
| | – Lợi nhuận ròng sau khi trả phí runner 0.30$/km | – Đo lường lợi nhuận thực trên từng đơn giao, cơ sở điều chỉnh phí giao hàng và mức lương runner (Finance, Accounting) | Finance (Tài chính)<br>Accounting (Kế toán) |



### **Mạch logic của toàn bộ dự án:**
- **A** là nền móng – cho biết hiện trạng kinh doanh cơ bản.  
- **B** là lá chắn – bảo vệ uy tín và giữ chân khách hàng thông qua chất lượng vận hành.  
- **C** là lưỡi dao tiết kiệm – cắt giảm chi phí nguyên liệu thừa, tận dụng cơ hội tăng doanh thu từ tùy chỉnh.  
- **D** là đòn bẩy tăng trưởng – tối ưu giá bán và đo lường lợi nhuận thực tế để tái đầu tư.
