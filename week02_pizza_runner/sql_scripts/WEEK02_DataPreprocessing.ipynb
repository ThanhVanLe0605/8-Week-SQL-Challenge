# GIAI ĐOẠN 3: KIỂM ĐỊNH CHẤT LƯỢNG DỮ LIỆU


## PHẦN 1.2: FRAMEWORK để xem xét vấn đề dữ liệu

**1. Schema và kiểu dữ liệu**

- Kiểm tra cột có tồn tại đúng tên, đúng thứ tự không.  
- Kiểm tra kiểu dữ liệu có phù hợp với bảng đích không  
  (ngày tháng phải là DATE/DATETIME, số phải là INT/DECIMAL, text đúng độ dài).  
- Xác định khóa chính thực tế  
  (composite key, surrogate key) để đảm bảo mỗi dòng là duy nhất.

---

**2. Tính đầy đủ (Completeness)**

- Xử lý các giá trị thể hiện “thiếu dữ liệu” như  
  NULL, 'null', 'NaN', '', ' ' → đồng nhất thành NULL.  
- Kiểm tra các cột bắt buộc (NOT NULL) không được phép thiếu dữ liệu.

---

**3. Tính duy nhất (Uniqueness)**

- Phát hiện bản ghi trùng lặp hoàn toàn hoặc theo business key.  
- Tạo surrogate key (dùng ROW_NUMBER()) cho các bảng thiếu khóa tự nhiên hoặc composite key quá dài.

---

**4. Chuẩn hóa và làm sạch (Standardization and Cleansing)**

- Trích xuất số từ chuỗi có đơn vị  
  (ví dụ '20km' → 20).  
- Tách cột có nhiều giá trị  
  (dạng '1,2,3') thành các dòng riêng (normalize).  
- Chuẩn hóa chữ hoa/thường  
  (ví dụ tên topping).  
- Xử lý các giá trị không hợp lệ  
  (khoảng trắng thừa, ký tự đặc biệt).

---

**5. Tính toàn vẹn tham chiếu (Referential Integrity)**

- Kiểm tra khóa ngoại có tồn tại trong bảng chính  
  (ví dụ pizza_id có trong pizza_names, runner_id có trong runners).

---

**6. Quy tắc nghiệp vụ (Business Rule Validation)**

- Xác định đơn nào được gửi dựa trên logic hủy (cancellation).  
- Kiểm tra giá trị hợp lệ theo ngữ cảnh:  
  distance > 0, duration > 0, order_date <= required_date, v.v.
```

NHẬN XÉT

 * Cần tiến hành phân rã chuỗi dữ liệu (*string splitting*) để chuyển đổi thuộc tính đa trị này thành các bản ghi độc lập trên nhiều dòng. Quá trình chuẩn hóa (*normalization*) này là bắt buộc để thiết lập mối quan hệ nhiều-nhiều (*Many-to-Many relationship*).
 * Toàn bộ công đoạn chuẩn hóa và làm sạch cấu trúc trên phải được hoàn tất trước khi tiến hành nạp dữ liệu (*import*) vào hệ thống bảng đích.


````sql
-- Tạo bảng tạm pizza_recipes_clean
CREATE TEMP TABLE pizza_recipes_clean(
	pizza_recipes_id BIGINT GENERATED ALWAYS AS IDENTITY,
	pizza_id INT,
	toppings INT
);
````

````sql
-- Xử lý và Insert trực tiếp 
---- Sử dụng SPLIT và EXPLODE để tách chuỗi toppings thành từng dòng riêng
SELECT *
FROM staging.pizza_recipes
LATERAL VIEW EXPLODE(SPLIT(toppings, ',')) t AS value ;
````

  
>  **QUAN SÁT VÀ NHẬN XÉT**

* **Ràng buộc thực thể :**
    * *Hiện trạng:* Vi phạm nguyên tắc xác định định danh duy nhất cho bản ghi.
    * *Giải pháp xử lý:* Định hình lại cấu trúc bảng bằng cách khởi tạo khóa thay thế (*Surrogate Key*) để đảm bảo tính toàn vẹn thực thể.

* **Chuẩn hóa miền giá trị và Kiểu dữ liệu :**
    * *Trường `pickup_time`:* Khắc phục tình trạng khuyết thiếu dữ liệu bằng cách loại bỏ các chuỗi text lỗi (`'null'`) và chuyển đổi đồng bộ về định dạng thời gian chuẩn (**TIMESTAMP**).

* **Trích xuất thuộc tính cấu trúc từ chuỗi văn bản :**
    * *Trường `distance`:* Dữ liệu không đồng nhất cấu trúc (tồn tại các biến thể chuỗi văn bản như `'20km'`, `'23.4 km'`, `'null'`). Tiến hành kỹ thuật tách chuỗi để loại bỏ đơn vị đơn vị đo lường, quy đổi chuỗi `'null'` về giá trị **NULL** hệ thống, và thực hiện ép kiểu số thực có độ chính xác cố định.
    * *Trường `duration`:* Tương tự, các giá trị đa dạng hình thức định dạng (`'32 minutes'`, `'20 mins'`, `'15 minute'`). Áp dụng giải pháp trích xuất phần số đơn thuần, xử lý các chuỗi trống thành định dạng số nguyên thích hợp.

* **Chuẩn hóa Logic nghiệp vụ (Business Logic Standardisation):**
    * *Trường `cancellation`:* Tồn tại các chuỗi rỗng (`''`), từ khóa văn bản lỗi (`'null'`) xen lẫn dữ liệu nghiệp vụ chính thức (`'Restaurant Cancellation'`, `'Customer Cancellation'`). Tiến hành quy đổi toàn bộ các định dạng không hợp lệ hoặc rỗng về trạng thái **NULL**, đồng thời bảo toàn nội dung văn bản gốc đối với các bản ghi ghi nhận hành vi hủy đơn thực tế phục vụ cho phân tích nghiệp vụ.

````sql
CREATE TEMP TABLE runner_orders_clean (
    runner_order_id    BIGINT GENERATED ALWAYS AS IDENTITY,
    order_id           INT,
    runner_id          INT,
    pickup_time        TIMESTAMP,
    distance           FLOAT,
    duration           INT,
    cancellation       STRING
);
````

````sql
-- Xử lý và INSERT vào clean 
INSERT INTO runner_orders_clean(order_id, runner_id, pickup_time , distance , duration , cancellation )
SELECT 
	 order_id,
	 runner_id,

	 -- Xử lý cột 'pickup_time': chuyển 'null'(chuỗi) thành NULL (giá trị), ép kiểu DATETIME
	 CASE 
		 WHEN pickup_time = 'null' THEN NULL
		 ELSE CAST(pickup_time AS TIMESTAMP)
	 END AS pickup_time,

	 -- Xử lý cột 'distance': xóa 'km', cắt khoảng trắng thừa, ép kiểu FLOAT để giữ số thập phân 
	 CASE
		WHEN distance = 'null' THEN NULL
		ELSE CAST(TRIM(REPLACE(distance,'km','')) AS FLOAT)
	 END AS distance,

	 -- Xử lý cột 'duration': xóa nhiều biến thể của chữ 'phút' như minutes, minute, mins -> cắt khoảng trắng -> ép kiểu INT 
	 CASE
		WHEN duration = 'null' THEN NULL
		ELSE  CAST(TRIM(REPLACE(REPLACE(REPLACE(duration,'minutes',''),'minute',''),'mins','')) AS INT)
	 END AS duration,

	 -- Xử lý cột 'cancellation': đưa 'null' (chuỗi) và ''(empty string) về chuẩn NULL 
	 CASE
		WHEN cancellation IS NULL OR cancellation = 'null' OR TRIM(cancellation) = '' THEN NULL
		ELSE cancellation 
	 END AS cancellation

FROM staging.runner_orders
````

````sql
-- kiểm tra lại
SELECT * 
FROM runner_orders_clean
````

>  **NHẬN XÉT**

Nhằm giải quyết bài toán trùng lặp logic nghiệp vụ và phân rã thuộc tính đa trị mà không làm mất đi các bản ghi hợp lệ, quy trình thực thi được chuẩn hóa qua mô hình kiến trúc Baseline như sau:

#### 1. Khởi tạo cấu trúc nền tảng 
* **Giải pháp định danh:** Không áp dụng từ khóa `DISTINCT` để khử trùng lặp trực tiếp, do cơ chế nghiệp vụ cho phép tồn tại nhiều sản phẩm giống cấu hình trong cùng đơn hàng. 
* **Kỹ thuật áp dụng:** Sử dụng BIGINT GENERATED ALWAYS AS IDENTITY. Khóa này đóng vai trò duy trì tính duy nhất cho từng thực thể bản ghi, giải quyết triệt để bài toán nhập nhằng khi tham chiếu logic ở các bước sau.
* **Xử lý hình thái học:** Đồng bộ hóa tất cả các ký tự khuyết thiếu hoặc lỗi biểu diễn (`'null'`, `NaN`, `''`) về trạng thái chuẩn hệ thống (**NULL**), đồng thời loại bỏ các khoảng trắng dư thừa sau ký hiệu phân tách chuỗi.

#### 2. Phân tách thực thể và Chuẩn hóa cấu trúc đa trị 
* **Kỹ thuật phân rã:** Sử dụng tiến trình tách chuỗi cấu trúc dữ liệu (`split` / `explode`) dựa trên định danh khóa `order_item_id` để phân rã hai trường `exclusions` và `extras` thành các token độc lập.
* **Ép kiểu và làm sạch miền giá trị:** * Loại bỏ các token rỗng phát sinh trong quá trình phân tách cấu trúc chuỗi lỗi.
    * Thực hiện ép kiểu dữ liệu của token về định dạng số nguyên (**INT**). Các token không thể định cấu hình (sai kiểu dữ liệu) sẽ bị loại bỏ khỏi tiến trình sạch.
* **Kiểm định toàn vẹn tham chiếu:** Đối chiếu hệ thống token sau khi trích xuất với danh mục gốc `staging.pizza_toppings`. Hệ thống chỉ chấp nhận và giữ lại các giá trị có `topping_id` tồn tại hợp lệ trong bảng danh mục.
 
#### 3. Định hình mô hình dữ liệu đích 
Sau khi chuẩn hóa bảng cơ sở , cấu trúc dữ liệu được phân tách vật lý thành 3 cấu trúc độc lập nhằm đảm bảo quy tắc chuẩn hóa cấu trúc:
* Bảng chính: `customer_orders_clean` (Đã loại bỏ các cột gây đa trị `exclusions` và `extras`, bảo toàn cấu trúc thuộc tính lõi).
* Bảng trung gian 1: `customer_orders_exclusions` (Liên kết trực tiếp với bảng danh mục toppings qua `topping_id`).
* Bảng trung gian 2: `customer_orders_extras` (Liên kết trực tiếp với bảng danh mục toppings qua `topping_id`).

````sql
-- Tạo bảng #customer_orders_base 
CREATE TEMP TABLE customer_orders_base(
		customer_orders_id BIGINT GENERATED ALWAYS AS IDENTITY,

		order_id INT,
		customer_id INT,
		pizza_id INT,

		exclusions STRING,	-- Tạm giữ lại text để bước sau xử lý
		extras STRING,		-- Tạm giữ lại text để bước sau xử lý

		order_time TIMESTAMP 
)
````

````sql
-- Làm sạch dữ liệu và INSERT vào bảng base
INSERT INTO customer_orders_base (order_id, customer_id, pizza_id, exclusions, extras,order_time)
SELECT
		order_id,
		customer_id,
		pizza_id,

		-- Chuẩn hóa exclusion
		CASE 
			WHEN exclusions IS NULL OR TRIM(exclusions) IN ('null', '') THEN NULL
			ELSE REPLACE(exclusions, ', ',',')
		END AS exclusions,

		-- Chuẩn hóa extras
		CASE
			WHEN extras IS NULL OR TRIM(extras) IN ('null', '') THEN NULL
			ELSE REPLACE(extras, ', ',',')
		END AS extras,

		order_time
FROM staging.customer_orders 
````

````sql
-- Tạo bảng chính (không còn cột đa trị)
CREATE TEMP TABLE customer_orders_clean(
	    customer_orders_id INT ,

		order_id INT,
		customer_id INT,
		pizza_id INT,
		
		order_time TIMESTAMP 
)
````

````sql
-- chèn dữ liệu từ bảng base vào bảng customer_orders_clean
INSERT INTO customer_orders_clean(customer_orders_id, order_id, customer_id, pizza_id, order_time)
SELECT
		customer_orders_id , 
		order_id ,
		customer_id ,
		pizza_id , 
		order_time
FROM customer_orders_base
````

````sql
-- Bảng 1: customer_orders_exclusions (temp table)
CREATE TEMP TABLE customer_orders_exclusions (
    customer_orders_id INT,
    topping_id INT
    
);

INSERT INTO customer_orders_exclusions (customer_orders_id, topping_id)
SELECT DISTINCT 
    base.customer_orders_id,
    TRY_CAST(TRIM(value) AS INT) AS topping_id
FROM customer_orders_base   base
LATERAL VIEW EXPLODE(SPLIT(base.exclusions, ',')) t AS value
WHERE 
    value != ''
    AND TRY_CAST(TRIM(value) AS INT) IS NOT NULL
    AND TRY_CAST(TRIM(value) AS INT) IN (SELECT topping_id FROM staging.pizza_toppings);

-- Bảng 2: customer_orders_extras (temp table)
CREATE TEMP TABLE customer_orders_extras (
    customer_orders_id INT,
    topping_id INT
);

INSERT INTO customer_orders_extras (customer_orders_id, topping_id)
SELECT DISTINCT 
    base.customer_orders_id,
    TRY_CAST(TRIM(value) AS INT) AS topping_id
FROM customer_orders_base base
LATERAL VIEW EXPLODE(SPLIT(base.extras, ',')) t AS value
WHERE 
    value != ''
    AND TRY_CAST(TRIM(value) AS INT) IS NOT NULL
    AND TRY_CAST(TRIM(value) AS INT) IN (SELECT topping_id FROM staging.pizza_toppings);
````

> * **Trạng thái khởi tạo:** Toàn bộ hệ thống cấu trúc bảng đích (*target tables*) hiện tại đều được thiết lập ở trạng thái rỗng (*blank tables*).
> * **Cơ chế kiểm soát trùng lặp:** Nhằm triệt tiêu rủi ro dư thừa dữ liệu phát sinh từ việc thực thi câu lệnh `INSERT INTO` lặp lại nhiều lần trên cùng một phiên làm việc (*session*), tiến trình cần tuân thủ nghiêm ngặt quy tắc giải phóng vùng nhớ.
> * **Giải pháp kỹ thuật:** Luôn luôn áp dụng lệnh `TRUNCATE TABLE` để làm sạch toàn bộ kho lưu trữ vật lý của bảng đích trước khi kích hoạt tiến trình nạp dữ liệu (*Data Loading*). Hành động này đảm bảo tính bất biến (*idempotency*) của pipeline – chạy lại nhiều lần vẫn cho ra một kết quả duy nhất.

### PHẦN 1.4.3: TẠO RÀNG BUỘC KHÓA NGOẠI

````sql
-- Thêm foreign key vào bảng destination.runner_orders  
		-- FOREIGN KEY: runner_id -> destination.runners(runner_id)
ALTER TABLE destination.runner_orders 
ADD CONSTRAINT FK_runners_runner_orders FOREIGN KEY (runner_id)
REFERENCES destination.runners(runner_id)
````

````sql

-- Thêm foreign key vào  bảng destination.pizza_recipes
		-- FOREIGN KEY: pizza_id -> destination.pizza_names(pizza_id)
		-- FOREIGN KEY: toppings -> destination.pizza_toppings(topping_id)
ALTER TABLE destination.pizza_recipes
ADD CONSTRAINT FK_pizza_names_recipes FOREIGN KEY (pizza_id)
REFERENCES destination.pizza_names(pizza_id);


ALTER TABLE destination.pizza_recipes
ADD CONSTRAINT FK_pizza_toppings_recipes FOREIGN KEY (toppings)
REFERENCES destination.pizza_toppings(topping_id)
````

````sql
-- Thêm foreign key vào bảng destination.customer_orders
		-- FOREIGN KEY: pizza_id -> destination.pizza_names(pizza_id)
ALTER TABLE destination.customer_orders
ADD CONSTRAINT FK_pizza_names_customer_orders FOREIGN KEY (pizza_id)
REFERENCES destination.pizza_names(pizza_id)
````

````sql
-- Thêm foreign key vào bảng destination.customer_orders_exclusions
		-- FOREIGN KEY: customer_orders_id -> destination.customer_orders(customer_orders_id)
		-- FOREIGN KEY: topping_id         -> destination.pizza_toppings(topping_id)
ALTER TABLE destination.customer_orders_exclusions
ADD CONSTRAINT FK_customer_orders_customer_orders_exclusions FOREIGN KEY (customer_orders_id)
REFERENCES destination.customer_orders(customer_orders_id);

ALTER TABLE destination.customer_orders_exclusions
ADD CONSTRAINT FK_pizza_toppings_customer_orders_exclusions FOREIGN KEY (topping_id)
REFERENCES destination.pizza_toppings(topping_id)
````

````sql
-- Thêm foreign key vào bảng destination.customer_orders_extras
		-- FOREIGN KEY: customer_orders_id -> destination.customer_orders(customer_orders_id)
		-- FOREIGN KEY: topping_id         -> destination.pizza_toppings(topping_id)
ALTER TABLE destination.customer_orders_extras
ADD CONSTRAINT FK_customer_orders_customer_orders_extras FOREIGN KEY (customer_orders_id)
REFERENCES destination.customer_orders(customer_orders_id);

ALTER TABLE destination.customer_orders_extras 
ADD CONSTRAINT FK_pizza_toppings_customer_orders_extras FOREIGN KEY (topping_id )
REFERENCES destination.pizza_toppings(topping_id)
````
